```python
image_np_original = cv2.imread("/media/test.jpg")
gray      = cv2.cvtColor(image_np_original, cv2.COLOR_BGR2GRAY)
h, w      = gray.shape

_, thresh = cv2.threshold(gray, 0, 255,
                          cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

y = 0
i = 0
k= h//9


while y + k <= h:
    strip   = thresh[y:y + k, :]
    roi_np  = image_np_original[y:y + k, :]
    output = model.predict(input=roi_np)
    for res in output:
      res.print()
    cv2.imwrite(f"/media/part{i}.jpg",roi_np)

    y += k
    i += 1

# Save leftover part at the bottom
if y < h:
    roi_leftover_np     = image_np_original[y:h, :]
    output = model.predict(input=roi_leftover_np)
    for res in output:
      res.print()
```

```python
from transformers import TrOCRProcessor, VisionEncoderDecoderModel
from PIL import Image
import cv2
import numpy as np
from transformers import TrOCRProcessor, VisionEncoderDecoderModel
from torch.utils.data import DataLoader
import torch
from torch.utils.data import Dataset
```

```python
final_img=Image.open("/media/test1.jpg")
```

Testing microsoft/trocr-base-handwritten

```python
processor = TrOCRProcessor.from_pretrained('microsoft/trocr-base-handwritten')
model = VisionEncoderDecoderModel.from_pretrained('microsoft/trocr-base-handwritten')
pixel_values = processor(images=final_img, return_tensors="pt").pixel_values

generated_ids = model.generate(pixel_values)
generated_text = processor.batch_decode(generated_ids, skip_special_tokens=True)[0]
```

```python
image_np_original = cv2.imread("/media/test.jpg")
gray      = cv2.cvtColor(image_np_original, cv2.COLOR_BGR2GRAY)
h, w      = gray.shape

_, thresh = cv2.threshold(gray, 0, 255,
                          cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

y = 0
i = 0
c= ''
k= h//9


while y + k <= h:
    strip   = thresh[y:y + k, :]
    roi_np  = image_np_original[y:y + k, :]

    pixel_values = processor(images=roi_np, return_tensors="pt").pixel_values
    model.eval()
    with torch.no_grad():
      generated_ids = model.generate(
      pixel_values,
      max_new_tokens=256  # ← tell it how long to generate: needs to be more
    )
    predicted_text = processor.batch_decode(generated_ids, skip_special_tokens=True)

    cv2.imwrite(f"/media/part{i}.jpg",roi_np)

    c += predicted_text[0]
    c += '\n'
    y += k
    i += 1

# Save leftover part at the bottom
if y < h:
    roi_leftover_np     = image_np_original[y:h, :]

    pixel_values = processor(images=roi_leftover_np, return_tensors="pt").pixel_values
    model.eval()
    with torch.no_grad():
      generated_ids = model.generate(
      pixel_values,
      max_new_tokens=256  # ← tell it how long to generate: needs to be more
    )
    predicted_text = processor.batch_decode(generated_ids, skip_special_tokens=True)

    c += predicted_text[0]
```
